```python
# STEP 1: 安裝 Clean-FID
# =======================
!pip install clean-fid

import zipfile
import os
from cleanfid import fid
```

```python
# STEP 2: 上傳 ZIP 檔案
# =======================
from google.colab import files

print("真實圖片集 zip 檔")
real_zip = files.upload()
real_zip_path = list(real_zip.keys())[0]

print("生成圖片集 zip 檔")
gen_zip = files.upload()
gen_zip_path = list(gen_zip.keys())[0]
```

```python
# STEP 3: 解壓縮
# =======================
real_dir = "/content/real_images"
gen_dir = "/content/generated_images"

os.makedirs(real_dir, exist_ok=True)
os.makedirs(gen_dir, exist_ok=True)

with zipfile.ZipFile(real_zip_path, 'r') as zip_ref:
    zip_ref.extractall(real_dir)

with zipfile.ZipFile(gen_zip_path, 'r') as zip_ref:
    zip_ref.extractall(gen_dir)

print(f"真實集解壓至: {real_dir}")
print(f"生成集解壓至: {gen_dir}")
```

```python
# STEP 4: 計算 Clean-FID
# =======================
fid_score = fid.compute_fid(real_dir, gen_dir, mode="clean")
print("✅ Clean-FID 分數:", fid_score)
```

```python
# STEP 5: 舊版對照
# =======================
fid_legacy = fid.compute_fid(real_dir, gen_dir, mode="legacy_pytorch")
print("ℹ️ Legacy PyTorch FID 分數:", fid_legacy)
```

```python
print("\n解讀建議：")
print("1) Clean-FID 是標準化流程，報告請用這個分數。")
print("2) 若 Legacy 與 Clean 差距很大，代表前處理方式會影響評估，需採用 Clean-FID。")
```
